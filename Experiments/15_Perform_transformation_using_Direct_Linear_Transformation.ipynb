```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("Screenshot 2026-02-26 140552.png")

if img is None:
    print("Image not loaded")
else:
    img = cv2.resize(img, (400, 300))

    rows, cols = img.shape[:2]

    src_pts = np.float32([
        [50, 50],
        [350, 50],
        [50, 250],
        [350, 250]
    ])

    dst_pts = np.float32([
        [20, 20],
        [380, 60],
        [60, 280],
        [340, 240]
    ])

    H, status = cv2.findHomography(src_pts, dst_pts)

    result = cv2.warpPerspective(img, H, (cols, rows))

    print("Original Image")
    cv2_imshow(img)

    print("DLT Transformed Image")
    cv2_imshow(result)
```
