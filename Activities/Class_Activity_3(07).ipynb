```python

#1.Shape Representation (Draw Shape Boundary)import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

cv2.drawContours(img, contours, -1, (0,255,0), 2)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
#3. Boundary Representation
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray,127,255,cv2.THRESH_BINARY)

contours,_ = cv2.findContours(thresh,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

boundary = img.copy()
cv2.drawContours(boundary,contours,-1,(255,0,0),2)

plt.imshow(cv2.cvtColor(boundary,cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
#5. Contour Hierarchy
import cv2

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png')
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

_,thresh = cv2.threshold(gray,127,255,cv2.THRESH_BINARY)

contours,hierarchy = cv2.findContours(thresh,cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)

print("Hierarchy:")
print(hierarchy)
```

```python
#6. Contour Area
import cv2

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png')
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

_,thresh = cv2.threshold(gray,127,255,cv2.THRESH_BINARY)

contours,_ = cv2.findContours(thresh,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

for c in contours:
    print("Area =",cv2.contourArea(c))
```

```python
#7. Contour Perimeter
import cv2

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png')
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

_,thresh = cv2.threshold(gray,127,255,cv2.THRESH_BINARY)

contours,_ = cv2.findContours(thresh,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

for c in contours:
    print("Perimeter =",cv2.arcLength(c,True))
```

```python
#8. Contour Orientation
import cv2
import numpy as np

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png')
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

_,thresh = cv2.threshold(gray,127,255,cv2.THRESH_BINARY)

contours,_ = cv2.findContours(thresh,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

for c in contours:
    if len(c)>=5:
        ellipse = cv2.fitEllipse(c)
        print("Orientation =",ellipse[2])
```

```python
#9. Contour Convexity
import cv2

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png')
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

_,thresh = cv2.threshold(gray,127,255,cv2.THRESH_BINARY)

contours,_ = cv2.findContours(thresh,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

for c in contours:
    print("Convex :",cv2.isContourConvex(c))
```

```python
#10. Canny Edge Detection
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png',0)

edges = cv2.Canny(img,100,200)

plt.imshow(edges,cmap='gray')
plt.title("Canny Edge")
plt.axis("off")
plt.show()
```

```python
#11. Sobel Operation
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png',0)

sobel = cv2.Sobel(img,cv2.CV_64F,1,1,ksize=3)

plt.imshow(sobel,cmap='gray')
plt.title("Sobel")
plt.axis("off")
plt.show()
```

```python
#12. Laplacian of Gaussian (LoG)
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png',0)

blur = cv2.GaussianBlur(img,(3,3),0)
log = cv2.Laplacian(blur,cv2.CV_64F)

plt.imshow(log,cmap='gray')
plt.title("Laplacian of Gaussian")
plt.axis("off")
plt.show()
```

```python
#13. Scharr Operation
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/Screenshot 2026-08-07 095258.png',0)

scharr = cv2.Scharr(img,cv2.CV_64F,1,0)

plt.imshow(scharr,cmap='gray')
plt.title("Scharr")
plt.axis("off")
plt.show()
```
