```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/images.jpg')

# Define Laplacian kernel with negative center coefficient
laplacian_kernel = np.array([[0, 1, 0],
                             [1, -4, 1],
                             [0, 1, 0]])

#  Apply the kernel to the color image
sharpened = cv2.filter2D(image, -1, laplacian_kernel)

cv2_imshow(image)
cv2_imshow(sharpened)
```
