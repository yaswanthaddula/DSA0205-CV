```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/images.jpg", 0)

# Boundary detection kernel
kernel = np.array([[-1, -1, -1],
                   [-1,  8, -1],
                   [-1, -1, -1]])

# Apply convolution
boundary = cv2.filter2D(img, -1, kernel)

cv2_imshow(boundary)
```
