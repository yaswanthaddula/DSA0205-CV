```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread("/content/images.jpg", 0)

# Create kernel
kernel = np.ones((5,5), np.uint8)

# Apply Black Hat
blackhat = cv2.morphologyEx(img, cv2.MORPH_BLACKHAT, kernel)

# Display result
cv2_imshow(blackhat)
```
