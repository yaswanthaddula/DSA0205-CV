```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('/content/images.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

# Gradient masks
gx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
gy = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

# Gradient magnitude
gradient = cv2.magnitude(gx, gy)
gradient = cv2.normalize(gradient, None, 0, 255, cv2.NORM_MINMAX).astype('uint8')

# Sharpen image using gradient mask
sharp = cv2.addWeighted(image, 1.0, cv2.cvtColor(gradient, cv2.COLOR_GRAY2RGB), 0.5, 0)

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sharp)
plt.title("Gradient Mask Sharpened")
plt.axis("off")

plt.show()
```
