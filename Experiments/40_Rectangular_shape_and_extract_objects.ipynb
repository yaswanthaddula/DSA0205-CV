```python
import cv2
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread("/content/Screenshot 2026-09-02 220041.png")

# Draw rectangle
cv2.rectangle(img, (50, 50), (300, 300), (0, 255, 0), 2)

# Extract object inside rectangle
object = img[50:300, 50:300]

# Display image with rectangle
cv2_imshow(img)

# Display extracted object
cv2_imshow(object)
```
